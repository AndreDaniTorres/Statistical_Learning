# **Perceptron Multicapa usando Keras**




Para mejorar el desempeño de la red, tenemos varias estrategias en diferentes etapas del proceso:

1. Datos:


*   Recolectar mas datos
*   Aumentar datos

2. Arquiterctura


*   Estrategias de inicializacion de pesos
*   Funciones de activacion

3. Normalizacion

*   Estandarizacion de datos de entrada
*   Normalizacion del batch

4. Entrenamiento

*   Tasa o paso de aprendizaje

5. Regularizacion


*   L2-L1
*   Early stopping
*   Dropout

```python
#Importacion de librerias
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
```

```python
#Carga de datos Fashion Mnist
fashion_mnist = keras.datasets.fashion_mnist
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()
#Normalizacion de datos
x_valid, x_train = x_train[:5000] / 255,x_train[5000:] / 255
y_valid, y_train = y_train[:5000], y_train[5000:]
x_test = x_test / 255
class_names = ["T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
               "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"]
```

## Regularizacion

importante para sobre o sub ajuste, viendo los datos de validación vs entrenamiento 

```python
#Modelo sin regularizacion

model = keras.models.Sequential([
    keras.layers.Flatten(input_shape=[28,28]),
    keras.layers.Dense(300, activation='relu'),
    keras.layers.Dense(100, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])

#Compilacion
optim = keras.optimizers.SGD(learning_rate=0.001)
model.compile(loss="sparse_categorical_crossentropy",
              optimizer= optim,
              metrics=["accuracy"])
#Entrenamiento
history = model.fit(x_train, y_train, epochs=15,
                    validation_data=(x_valid, y_valid))
#Evaluacion
model.evaluate(x_test, y_test)
```

termino adicional

regularizar el tamaño de los pesos
$\lambda$ que tanto influye

## L2

$ \frac{\lambda}{n} \sum^L_1 ||w||_2^2$

no introduce funciones no derivables 

```python
#Modelo con regularizacion l2
model = keras.models.Sequential([
    keras.layers.Flatten(input_shape=[28,28]),
    keras.layers.Dense(300, activation='relu', kernel_regularizer=keras.regularizers.l2(0.01)), #parámetro lambda = 0.01
    keras.layers.Dense(100, activation='relu', kernel_regularizer=keras.regularizers.l2(0.01)),
    keras.layers.Dense(10, activation='softmax', kernel_regularizer=keras.regularizers.l2(0.01))
])
#Compilacion
optim = keras.optimizers.SGD(learning_rate=0.001)
model.compile(loss="sparse_categorical_crossentropy",
              optimizer= optim,
              metrics=["accuracy"])
#Entrenamiento
history = model.fit(x_train, y_train, epochs=15,
                    validation_data=(x_valid, y_valid))
#Evaluacion
model.evaluate(x_test, y_test)
```

## L1

$ \frac{\lambda}{n} \sum^L_1 ||w||$

es mas fuerte: quedan pesos iguales a cero, es decir, apaga las neuronas 

```python
#Modelo con regularizacion l1
model = keras.models.Sequential([
    keras.layers.Flatten(input_shape=[28,28]),
    keras.layers.Dense(300, activation='relu', kernel_regularizer=keras.regularizers.l1(0.01)), #parámetro lambda = 0.01
    keras.layers.Dense(100, activation='relu', kernel_regularizer=keras.regularizers.l1(0.01)),
    keras.layers.Dense(10, activation='softmax', kernel_regularizer=keras.regularizers.l1(0.01))
])
#Compilacion
optim = keras.optimizers.SGD(learning_rate=0.001)
model.compile(loss="sparse_categorical_crossentropy",
              optimizer= optim,
              metrics=["accuracy"])
#Entrenamiento
history = model.fit(x_train, y_train, epochs=15,
                    validation_data=(x_valid, y_valid))
#Evaluacion
model.evaluate(x_test, y_test)
```

## L1 - L2

```python
#Modelo con regularizacion l2-l1
model = keras.models.Sequential([
    keras.layers.Flatten(input_shape=[28,28]),
    keras.layers.Dense(300, activation='relu', kernel_regularizer=keras.regularizers.l1_l2(l1=0.001, l2=0.01)), #parámetro lambda = 0.01
    keras.layers.Dense(100, activation='relu', kernel_regularizer=keras.regularizers.l1_l2(l1=0.001, l2=0.01)),
    keras.layers.Dense(10, activation='softmax', kernel_regularizer=keras.regularizers.l1_l2(l1=0.001, l2=0.01))
])
#Compilacion
optim = keras.optimizers.SGD(learning_rate=0.001)
model.compile(loss="sparse_categorical_crossentropy",
              optimizer= optim,
              metrics=["accuracy"])
#Entrenamiento
history = model.fit(x_train, y_train, epochs=15,
                    validation_data=(x_valid, y_valid))
#Evaluacion
model.evaluate(x_test, y_test)
```

## Optimizadores

```python
#Construccion del MLP

model = keras.models.Sequential([
    keras.layers.Flatten(input_shape=[28,28]),
    keras.layers.Dense(300, activation='relu'),
    keras.layers.Dense(100, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])

#Compilacion
optim = keras.optimizers.SGD(learning_rate=0.001, momentum=0.9)  # gradiente se mueve más rápido hacia el mínimo 
model.compile(loss="sparse_categorical_crossentropy",
              optimizer= optim,
              metrics=["accuracy"])
#Entrenamiento
history = model.fit(x_train, y_train, epochs=15,
                    validation_data=(x_valid, y_valid))
#Evaluacion
model.evaluate(x_test, y_test)
```

## adam

RMSProp + momentum

```python
#Construccion del MLP

model = keras.models.Sequential([
    keras.layers.Flatten(input_shape=[28,28]),
    keras.layers.Dense(300, activation='relu'),
    keras.layers.Dense(100, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])

#Compilacion
optim = keras.optimizers.Adam(learning_rate=0.001)  # gradiente se mueve más rápido hacia el mínimo
model.compile(loss="sparse_categorical_crossentropy",
              optimizer= optim,
              metrics=["accuracy"])
#Entrenamiento
history = model.fit(x_train, y_train, epochs=15,
                    validation_data=(x_valid, y_valid))
#Evaluacion
model.evaluate(x_test, y_test)
```

## Batch Normalization

Normalizar en cada capa

Mas costoso computacionalmente

```python
#Construccion del MLP

model = keras.models.Sequential([
    keras.layers.Flatten(input_shape=[28,28]),
    keras.layers.Dense(300, activation='relu'),
    keras.layers.BatchNormalization(),
    keras.layers.Dense(100, activation='relu'),
    keras.layers.BatchNormalization(),
    keras.layers.Dense(10, activation='softmax')
])

#Compilacion
optim = keras.optimizers.Adam(learning_rate=0.001)
model.compile(loss="sparse_categorical_crossentropy",
              optimizer= optim,
              metrics=["accuracy"])
#Entrenamiento
history = model.fit(x_train, y_train, epochs=15,
                    validation_data=(x_valid, y_valid))
#Evaluacion
model.evaluate(x_test, y_test)
```
