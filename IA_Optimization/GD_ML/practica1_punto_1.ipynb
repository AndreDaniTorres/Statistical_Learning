Encuentre el valor mínimo de las siguientes funciones empleando la búsqueda por grilla y la aleatoria. Para cada una de las funciones, muestre su gráfica e imprima el valor óptimo.


```python
import numpy as np
import matplotlib.pyplot as plt
```

## Punto 1

1. $𝑓(𝑥,𝑦) = 0.26(𝑥^{2}+𝑦^{2}−0.48𝑥𝑦)$ para $𝑥,𝑦 ∈ [−10,10]$


```python
def funcion1(x, y):
    return 0.26* ( x**2 + y**2 - 0.48*x*y )
```

```python
def plot(funcion, xy_lim, step):
    y = np.arange(xy_lim[0], xy_lim[1], step)
    x = np.arange(xy_lim[0], xy_lim[1], step)
    X, Y = np.meshgrid(x, y)
    Z = funcion(X, Y)
    print('Z shape:' ,Z.shape)
    
    # crear una imagen en 3D
    fig = plt.figure()
    ax = fig.add_subplot(111, projection='3d')
    ax.plot_surface(X, Y, Z, cmap='jet')
    plt.show()
```

```python
plot(funcion1, xy_lim=(-10, 10), step=0.1)
```

### Búsqueda

```python
def search(funcion1, x, y, x_sample, y_sample):
    
    X_sample, Y_sample = np.meshgrid(x_sample, y_sample)
    Z_sample = funcion1(X_sample, Y_sample)
    print('Z shape',Z_sample.shape)

    #Busqueda en grilla
    f_opt = Z_sample[0,0]
    best_x = 0
    best_y = 0

    # recorrer filas
    for i in range(Z_sample.shape[0]):
        # recorrer columnas
        for j in range(Z_sample.shape[1]):
            if Z_sample[i, j] < f_opt:
                f_opt = Z_sample[i, j]
                best_x = i
                best_y = j
    
    print('Punto optimo: x =',X_sample[best_x][best_y],'y=',Y_sample[best_x][best_y])
    print('Valor optimo f(x,y) = ' ,f_opt)
    
    
    #Grafica
    X, Y = np.meshgrid(x, y)
    Z = funcion1(X, Y)
    plt.contourf(x, y, Z, levels=50, cmap='jet')
    plt.scatter(X_sample, Y_sample, c='black', marker='.')
    plt.scatter(X_sample[best_x][best_y], Y_sample[best_x][best_y], c='red', marker='.')
    plt.colorbar()
    plt.show()
                
    return best_x, best_y, f_opt
```

### Búsqueda en grilla

```python
r_min, r_max = -10, 10
x1= np.arange(r_min, r_max, 0.1)
y1= np.arange(r_min, r_max, 0.1)
x1_sample = np.linspace(-3, 3, 20)
y1_sample = np.linspace(-3, 3, 20)

best_x, best_y, f_opt = search(funcion1, x=x1, y=y1, x_sample=x1_sample, y_sample=y1_sample)
```

### Búsqueda aleatoria

```python
r_min, r_max = -10, 10

x2= np.arange(r_min, r_max, 0.1)
y2= np.arange(r_min, r_max, 0.1)
x2_sample = r_min + np.random.rand(50) * (r_max - r_min)
y2_sample = r_min + np.random.rand(50) * (r_max - r_min)

best_x, best_y, f_opt = search(funcion1, x=x2, y=y2, x_sample=x2_sample, y_sample=y2_sample)
```

## Punto 2

2. $𝑓(𝑥,𝑦) = −cos(𝑥)cos(𝑦)e^{−((𝑥−𝜋)2+(𝑦−𝜋)2)}$ para $𝑥,𝑦 ∈ [−10,10]$

```python
def funcion2(x, y):
    return -np.cos(x) * np.cos(y) * np.exp(-((x - np.pi)**2 + (y - np.pi)**2))
```

```python
plot(funcion2, xy_lim=(-10, 10), step=0.1)
```

### Búsqueda en grilla 

```python
r_min, r_max = -10, 10
x3= np.arange(r_min, r_max, 0.1)
y3= np.arange(r_min, r_max, 0.1)
x3_sample = np.linspace(0, 5, 15)
y3_sample = np.linspace(0, 5, 15)

best_x, best_y, f_opt = search(funcion2, x=x3, y=y3, x_sample=x3_sample, y_sample=y3_sample)
```

### Búsqueda aleatoria

```python
r_min, r_max = -10, 10

x2= np.arange(r_min, r_max, 0.1)
y2= np.arange(r_min, r_max, 0.1)
x2_sample = r_min + np.random.rand(50) * (r_max - r_min)
y2_sample = r_min + np.random.rand(50) * (r_max - r_min)

best_x, best_y, f_opt = search(funcion2, x=x2, y=y2, x_sample=x2_sample, y_sample=y2_sample)
```
